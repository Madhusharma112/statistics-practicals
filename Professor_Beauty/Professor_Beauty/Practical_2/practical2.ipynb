```python
# PROFESSOR EVALUATION AND BEAUTY DATASET

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# LOAD DATASET
df = pd.read_csv("C:/Users/Ajsha/Downloads/evals.csv")

print("Dataset loaded successfully!")
print("Number of observations:", len(df))

print("\nFirst 5 rows:")
print(df.head())
```

```python
# Q1. TIME SERIES OR CROSS-SECTIONAL?

print("\nQ1. TYPE OF DATA")
print("The teacher rating data is CROSS-SECTIONAL data.")
print("Reason: The dataset contains observations of different")
print("professors/courses and is not collected over time.")

# Q2. MEAN, MEDIAN, MINIMUM AND MAXIMUM OF STUDENTS

print("\nQ2. NUMBER OF STUDENTS")

print("Mean    :", round(df["cls_students"].mean(), 2))
print("Median  :", df["cls_students"].median())
print("Minimum :", df["cls_students"].min())
print("Maximum :", df["cls_students"].max())

# Q3. DESCRIPTIVE STATISTICS
print("\nQ3. DESCRIPTIVE STATISTICS")

print(df.describe())

# Q4. HISTOGRAM OF BEAUTY VARIABLE

print("\nQ4. HISTOGRAM OF BEAUTY SCORE")

plt.hist(
    df["bty_avg"],
    bins=10,
    color="purple",
    edgecolor="black"
)

plt.xlabel("Beauty Score")
plt.ylabel("Frequency")
plt.title("Histogram of Professor Beauty Scores")

plt.show()

print("Comment:")
print("The beauty score distribution is approximately unimodal.")
print("Most observations are concentrated around the middle range.")

# Q5. BEAUTY SCORE BY GENDER

print("\nQ5. BEAUTY SCORE BY GENDER")
result = df.groupby("gender")["bty_avg"].agg(["mean", "std"])
print(result.round(3))
```

```python
# Q6. PERCENTAGE OF MALES AND FEMALES WHO ARE TENURED

print("\nQ6. TENURE STATUS BY GENDER")

# Total male and female professors
male_total = (df["gender"] == "male").sum()
female_total = (df["gender"] == "female").sum()

# Number of tenured male and female professors
male_tenured = ((df["gender"] == "male") &
                (df["rank"] == "tenured")).sum()

female_tenured = ((df["gender"] == "female") &
                  (df["rank"] == "tenured")).sum()

# Calculate percentages
male_percentage = (male_tenured / male_total) * 100
female_percentage = (female_tenured / female_total) * 100

print("Percentage of Male Professors who are Tenured:",
      round(male_percentage, 2), "%")

print("Percentage of Female Professors who are Tenured:",
      round(female_percentage, 2), "%")

# Conclusion
if male_percentage != female_percentage:
    print("Tenure status differs by gender.")
else:
    print("Tenure status does not differ by gender.")
```
