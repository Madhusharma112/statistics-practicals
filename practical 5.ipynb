```python
# Q1. Probability of evaluation score greater than 4.5

import pandas as pd

df = pd.read_csv("C:/Users/Ajsha/Downloads/teachingratings.csv")
print(df.head())
```

```python
# Q1. Probability of evaluation score greater than 4.5

import math

eval = df["eval"]
mean = eval.mean()
std = eval.std()

z = (4.5 - mean) / std
probability = 0.5 * (1 - math.erf(z / math.sqrt(2)))

print("Probability:", round(probability * 100, 1), "%")
```

```python
# Q2. Probability of evaluation score between 3.5 and 4.2

z1 = (3.5 - mean) / std
z2 = (4.2 - mean) / std

probability = 0.5 * (math.erf(z2 / math.sqrt(2)) -
                     math.erf(z1 / math.sqrt(2)))

print("Probability:", round(probability * 100, 1), "%")
```

```python
# Q3. Two-tailed test using normal distribution

import math

mu = 12
sigma = 5.5
x_bar = 10.7
n = 36

# Calculate Z-value
z = (x_bar - mu) / (sigma / math.sqrt(n))

# Calculate two-tailed p-value
p_value = 2 * (1 - 0.5 * (1 + math.erf(abs(z) / math.sqrt(2))))

print("Z-value:", round(z, 2))
print("P-value:", round(p_value, 4))
```

```python
# Q4. State the Null and Alternative Hypothesis

# Null Hypothesis
print("H0: μ = 12")

# Alternative Hypothesis
print("H1: μ ≠ 12")


# Graph for Two-Tailed Test
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu = 12
sigma = 5.5
n = 36
x_bar = 10.7

se = sigma / np.sqrt(n)

x = np.linspace(8, 16, 500)
y = norm.pdf(x, mu, se)

plt.plot(x, y)

plt.axvline(mu, linestyle='--', label='Historic Mean = 12')
plt.axvline(x_bar, linestyle='--', label='Sample Mean = 10.7')

plt.xlabel("Mean Score")
plt.ylabel("Density")
plt.title("Two-Tailed Hypothesis Test")
plt.legend()
plt.grid()

plt.show()
```
