```python
import pandas as pd
import numpy as np
from scipy import stats

# Load the dataset
df = pd.read_csv("C:/Users/Ajsha/Downloads/teachingratings.csv")

# Display first 5 rows
df.head()
```

```python
# Q1. T-Test: Using the teachers' rating data set, does gender affect teaching evaluation rates?

# Separate teaching evaluation scores by gender
male_eval = df.loc[df['gender'] == 'male', 'eval']
female_eval = df.loc[df['gender'] == 'female', 'eval']

# Perform independent two-sample Welch's t-test
t_stat, p_value = stats.ttest_ind(
    male_eval,
    female_eval,
    equal_var=False
)

print("T-statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 6))
```

```python
# Q2. ANOVA: Does beauty score differ by age?

bins = [0, 40, 50, 60, float('inf')]
labels = ['≤40', '41-50', '51-60', '61+']

df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)

groups = [df[df['age_group'] == g]['beauty'].dropna() for g in labels]

f_stat, p_anova = stats.f_oneway(*groups)

print("F-statistic =", round(f_stat, 3))
print("p-value =", f"{p_anova:.2e}")
```

```python
# Q3. Chi-Square: Is tenure associated with gender?

table = pd.crosstab(df['tenure'], df['gender'])

chi2, p, dof, expected = stats.chi2_contingency(table)

print("Chi-square =", round(chi2, 4))
print("df =", dof)
print("p-value =", round(p, 6))
```

```python
# Q4. Correlation: Using the teachers' rating dataset, is teaching evaluation score correlated with beauty score?

# Calculate Pearson correlation
r, p_corr = stats.pearsonr(df['eval'], df['beauty'])

# Display results
print("Pearson's r =", round(r, 3))
print("p-value =", f"{p_corr:.2e}")
```
