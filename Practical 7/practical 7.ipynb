```python
import pandas as pd
import statsmodels.api as sm

# Load dataset
df = pd.read_csv("C:/Users/Ajsha/Downloads/teachingratings.csv")

# Q1. Regression with T-test
# Does gender affect teaching evaluation rates?

# Independent variable
X = df["female"]
X = sm.add_constant(X)

# Dependent variable
y = df["eval"]

# Fit regression model
model = sm.OLS(y, X).fit()

# Regression output
print(model.summary())

# Interpretation
p_value = model.pvalues["female"]
coef = model.params["female"]

print("\nInterpretation:")
print("The p-value is 0.001, which is less than 0.05.")
print("Therefore, we reject the null hypothesis.")
print("Gender has a statistically significant effect on teaching evaluation scores.")
print("The coefficient of female is -0.168, indicating that the female category")
print("has an average evaluation score about 0.168 points lower than the reference category.")
print("Conclusion: Gender significantly affects teaching evaluation scores.")
```

```python
import statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm

# Q2. Regression with ANOVA
# Does beauty score for instructors differ by age?

# Create age groups
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 40, 56, float("inf")],
    labels=["Young", "Middle", "Old"]
)

# Regression model
model = smf.ols("beauty ~ C(age_group)", data=df).fit()

# ANOVA table
anova_result = anova_lm(model)

# Rename the row
anova_result.index = ["age_group", "Residual"]

# Display only age_group row
print(anova_result.loc[["age_group"]])

# Get p-value
p_value = anova_result.loc["age_group", "PR(>F)"]

# Interpretation
print("\nInterpretation:")
print(f"The p-value is {p_value:.2e}, which is less than 0.05.")
print("Therefore, we reject the null hypothesis.")
print("There is a statistically significant difference in beauty scores among the age groups.")
print("Conclusion: Beauty score significantly differs by age group.")
```

```python
# Q3. Correlation
# Is teaching evaluation score correlated with beauty score?

# Independent variable
X = df["beauty"]

# Add constant
X = sm.add_constant(X)

# Dependent variable
y = df["eval"]

# Fit OLS regression model
model = sm.OLS(y, X).fit()

# Display regression output
print(model.summary())

# Get values for interpretation
p_value = model.pvalues["beauty"]
coef = model.params["beauty"]
r_squared = model.rsquared

# Interpretation
print("\nInterpretation:")
print(f"The p-value is {p_value:.2e}, which is less than 0.05.")
print("Therefore, we reject the null hypothesis.")
print("Beauty score is significantly associated with teaching evaluation score.")
print(f"The coefficient of beauty is {coef:.3f}, indicating a positive association.")
print(f"The R-squared value is {r_squared:.3f}, meaning the model explains approximately {r_squared*100:.1f}% of the variation in evaluation scores.")
print("Conclusion: Teaching evaluation score is significantly and positively associated with beauty score.")
```
