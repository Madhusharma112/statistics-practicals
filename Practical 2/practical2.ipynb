```python
# PROFESSOR EVALUATION AND BEAUTY DATASET

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# LOAD DATASET
df = pd.read_csv("C:/Users/Ajsha/Downloads/ProfEvaltnsBeautyPublic.csv")

print("Dataset loaded successfully!")
print("Number of observations:", len(df))

print("\nFirst 5 rows:")
print(df.head())
```

```python
# Display the shape of the dataset
print("Number of observations:", df.shape[0])
print("Number of variables:", df.shape[1])

# Display the first few rows
df.head()
print("Interpretation:The professor rating data is cross-sectional data because it contains observations of different professors/instructors at a particular point/period rather than observations collected over time.")
```

```python
# Q2: Mean, Median, Minimum and Maximum

print("Mean:", round(df["students"].mean(), 2))
print("Median:", round(df["students"].median(), 2))
print("Minimum:", df["students"].min())
print("Maximum:", df["students"].max())
```

```python
# Q3: Descriptive Statistics

descriptive_stats = df.describe()

print("Descriptive Statistics Table:")
descriptive_stats
```

```python
# Q4: Histogram of Beauty Variable

plt.hist(df["btystdave"], bins=15, edgecolor="black")

plt.xlabel("Beauty Score")
plt.ylabel("Frequency")
plt.title("Histogram of Beauty Score")

plt.show()
```

```python
# Q5: Average Beauty Score by Gender

gender = df.groupby("female")["btystdave"].agg(["mean", "std"])

gender.index = ["Male", "Female"]

print(gender)
```

```python
# Percentage of tenured professors by gender

tenured_percentage = df.groupby("female")["tenured"].mean() * 100

tenured_percentage.index = ["Male", "Female"]

print(tenured_percentage)
```
